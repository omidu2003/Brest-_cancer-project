## 📊 Correlation Matrix of Numerical Features

This section computes and visualizes the correlation matrix for all columns that can be converted to numeric data types. Non-numeric columns (e.g., strings or categories) are automatically excluded.

- Columns are first converted to `float` where possible.
- A heatmap is generated to highlight the strength and direction of correlations.
- Only meaningful numerical relationships are visualized.


```python
def plot_correlation_matrix(df):
    """
    Converts columns to float where possible and plots the correlation matrix for numeric columns.

    Parameters:
    df (pd.DataFrame): Input DataFrame.

    Returns:
    pd.DataFrame: The computed correlation matrix.
    """
    df = df.copy()
    numeric_columns = []

    # Attempt to convert columns to float and collect numeric ones
    for column in df.columns:
        try:
            df[column] = df[column].astype(float)
            numeric_columns.append(column)
        except (ValueError, TypeError):
            continue  # Skip non-numeric columns

    # Subset only numeric columns
    numeric_df = df[numeric_columns]

    # Compute correlation matrix
    correlation_matrix = numeric_df.corr()

    # Plot the heatmap
    plt.figure(figsize=(16, 12))
    sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
    plt.title("Correlation Matrix (Numerical Columns)")
    plt.tight_layout()
    plt.show()

    return correlation_matrix
```

## Cluster Analysis

### K-Means cluster Analysis

We apply K-Means clustering on the numerical features (excluding ID and target)
to discover natural groupings in the data, then compare them with the actual
diagnosis labels.


```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

# 1️⃣ Select features for clustering (numeric, no ID, no target)
features = df.select_dtypes(include=['int64', 'float64']).copy()
features = features.drop(columns=['id'], errors='ignore')  # drop id if present

# Do NOT use the target column for clustering
if 'diagnosis' in df.columns:
    # diagnosis is usually object type, but just in case:
    features = features.loc[:, [col for col in features.columns if col != 'diagnosis']]

# 2️⃣ Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(features)

# 3️⃣ Choose k using the Elbow Method
wcss = []  # within-cluster sum of squares
K = range(2, 11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(6, 4))
plt.plot(K, wcss, marker='o')
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('WCSS')
plt.show()

# 4️⃣ Fit final K-Means model (here we choose k=2 for benign vs malignant)
k_optimal = 2
kmeans = KMeans(n_clusters=k_optimal, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_scaled)

# Add cluster labels to the original dataframe
df['cluster'] = cluster_labels

# 5️⃣ Evaluate clustering with Silhouette Score
sil_score = silhouette_score(X_scaled, cluster_labels)
print(f"Silhouette Score for k={k_optimal}: {sil_score:.4f}")

# 6️⃣ Visualize clusters in 2D using PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(6, 4))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=cluster_labels, alpha=0.7)
plt.title('K-Means Clusters (PCA-reduced features)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()

# 7️⃣ Compare clusters with actual diagnosis labels (if available)
if 'diagnosis' in df.columns:
    print("\nCluster vs Diagnosis Crosstab:")
    print(pd.crosstab(df['cluster'], df['diagnosis']))
```

### Agglomerative Clustering

We apply Agglomerative (Hierarchical) Clustering on the numerical features to
discover natural groupings in the data and compare them with the diagnosis labels.

```python
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import dendrogram, linkage

# Select features for clustering (numeric, exclude ID and target if present)
features = df.select_dtypes(include=['int64', 'float64']).copy()
features = features.drop(columns=['id'], errors='ignore')  # drop id if it exists

# Do NOT use the target column for clustering
if 'diagnosis' in df.columns and 'diagnosis' in features.columns:
    features = features.drop(columns=['diagnosis'])

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(features)

# Agglomerative Clustering (choose number of clusters, e.g., 2)
agg_model = AgglomerativeClustering(
    n_clusters=2,
    metric='euclidean',
    linkage='ward'
)


cluster_labels_agg = agg_model.fit_predict(X_scaled)

# Add cluster labels to the dataframe
df['agg_cluster'] = cluster_labels_agg

# Silhouette Score to evaluate clustering quality
sil_score_agg = silhouette_score(X_scaled, cluster_labels_agg)
print(f"Silhouette Score (Agglomerative, k=2): {sil_score_agg:.4f}")

# Visualize clusters in 2D using PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(6, 4))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=cluster_labels_agg, alpha=0.7)
plt.title('Agglomerative Clustering (PCA-reduced features)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()

# Dendrogram (on a subset of points to keep it readable)
sample_size = 80  # use a subset to avoid a messy dendrogram
X_sample = X_scaled[:sample_size]

linked = linkage(X_sample, method='ward')

plt.figure(figsize=(10, 5))
dendrogram(linked, truncate_mode='level', p=5)
plt.title('Hierarchical Clustering Dendrogram (Sample)')
plt.xlabel('Sample index')
plt.ylabel('Distance')
plt.show()

# Compare clusters with actual diagnosis labels (if available)
if 'diagnosis' in df.columns:
    print("\nAgglomerative Cluster vs Diagnosis Crosstab:")
    print(pd.crosstab(df['agg_cluster'], df['diagnosis']))
```

### Models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier, ExtraTreesClassifier
from xgboost import XGBClassifier


# Class names for 0 and 1
target_names = ['Benign', 'Malignant']   # 0 -> Benign, 1 -> Malignant

# 1️⃣ Define the models
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "SVM (RBF kernel)": SVC(kernel='rbf', probability=True, random_state=42),
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42
    ),
    "K-Nearest Neighbors": KNeighborsClassifier(n_neighbors=5),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "AdaBoost": AdaBoostClassifier(random_state=42),
    "Extra Trees": ExtraTreesClassifier(
        n_estimators=200,
        random_state=42
    ),
      "XGBoost": XGBClassifier(
        n_estimators=200,
        learning_rate=0.1,
        max_depth=3,
        subsample=1.0,
        colsample_bytree=1.0,
        random_state=42,
        eval_metric="logloss"
    ),
}
```

### Model Training & Validation

```python
from sklearn.metrics import roc_auc_score, roc_curve


def train_and_evaluate_models(X_train_scaled, X_test_scaled):
    # 2️⃣ Train and evaluate each model
    for name, model in models.items():
        print("=" * 70)
        print(f"Model: {name}")
        print("=" * 70)

        # Train
        model.fit(X_train_scaled, y_train)

        # Predict
        y_pred = model.predict(X_test_scaled)

        # Classification report
        print("\nClassification Report:")
        print(classification_report(y_test, y_pred, target_names=target_names))

        # Confusion matrix (raw)
        cm = confusion_matrix(y_test, y_pred)

        # Confusion matrix as labeled table
        cm_df = pd.DataFrame(
            cm,
            index=[f"Actual {c}" for c in target_names],
            columns=[f"Pred {c}" for c in target_names]
        )

        print("\nConfusion Matrix (labeled):")
        print(cm_df)

        # 🔹 NEW: AUC-ROC
        # Get scores/probabilities for positive class
        if hasattr(model, "predict_proba"):
            y_scores = model.predict_proba(X_test_scaled)[:, 1]
        elif hasattr(model, "decision_function"):
            y_scores = model.decision_function(X_test_scaled)
        else:
            y_scores = None

        if y_scores is not None:
            auc = roc_auc_score(y_test, y_scores)
            print(f"\nAUC-ROC: {auc:.4f}")

            # Optional: ROC curve plot
            fpr, tpr, _ = roc_curve(y_test, y_scores)
            plt.figure(figsize=(5, 4))
            plt.plot(fpr, tpr, label=f"AUC = {auc:.3f}")
            plt.plot([0, 1], [0, 1], 'k--')
            plt.title(f"ROC Curve - {name}")
            plt.xlabel("False Positive Rate")
            plt.ylabel("True Positive Rate")
            plt.legend(loc="lower right")
            plt.tight_layout()
            plt.show()

        # 🔹 NEW: Seaborn heatmap for confusion matrix
        plt.figure(figsize=(5, 4))
        sns.heatmap(cm_df, annot=True, fmt='d', cmap='Blues')
        plt.title(f"Confusion Matrix - {name}")
        plt.ylabel("Actual")
        plt.xlabel("Predicted")
        plt.tight_layout()
        plt.show()

        print("\n\n")


```

Model training with Standard scalling Data

```python
# Call the function like this:
train_and_evaluate_models(X_train_scaled_Standard, X_test_scaled_Standard)
```

Model training with Min-Max scalling Data

```python
train_and_evaluate_models(X_train_scaled_MinMax, X_test_scaled_MinMax)
```

## Hyperparameter Tuning

* Use the Randomizd search for the hyper parameter tuning

### Hyperparameters Tuning for the RandomForest

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve  # NEW

# Random Forest hyperparameter tuning
rf_base = RandomForestClassifier(random_state=42)

rf_param_dist = {
    "n_estimators": [100, 200, 300, 500],
    "max_depth": [None, 5, 10, 15, 20],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "max_features": ["sqrt", "log2", 0.5, 0.7],
}

rf_random_search = RandomizedSearchCV(
    estimator=rf_base,
    param_distributions=rf_param_dist,
    n_iter=20,
    scoring="f1",
    cv=5,
    n_jobs=-1,
    random_state=42,
    verbose=1,
)

rf_random_search.fit(X_train_scaled_Standard, y_train)

print("Best Random Forest params:")
print(rf_random_search.best_params_)

best_rf = rf_random_search.best_estimator_

# Evaluate on test set
y_pred_rf = best_rf.predict(X_test_scaled_Standard)

print("\n=== Tuned Random Forest - Classification Report ===")
print(classification_report(y_test, y_pred_rf, target_names=target_names))

cm_rf = confusion_matrix(y_test, y_pred_rf)
cm_rf_df = pd.DataFrame(
    cm_rf,
    index=[f"Actual {c}" for c in target_names],
    columns=[f"Pred {c}" for c in target_names]
)

print("\nConfusion Matrix (Random Forest, labeled):")
print(cm_rf_df)

plt.figure(figsize=(5, 4))
sns.heatmap(cm_rf_df, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix - Tuned Random Forest")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.tight_layout()
plt.show()

# NEW: AUC-ROC + ROC curve for tuned Random Forest
y_scores_rf = best_rf.predict_proba(X_test_scaled_Standard)[:, 1]  # probabilities for class 1
auc_rf = roc_auc_score(y_test, y_scores_rf)
print(f"\nAUC-ROC (Tuned Random Forest): {auc_rf:.4f}")

fpr_rf, tpr_rf, _ = roc_curve(y_test, y_scores_rf)

plt.figure(figsize=(5, 4))
plt.plot(fpr_rf, tpr_rf, label=f"AUC = {auc_rf:.3f}")
plt.plot([0, 1], [0, 1], 'k--')
plt.title("ROC Curve - Tuned Random Forest")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

### Hyperparameters Tuning for the Xgboost

```python
from xgboost import XGBClassifier

# ----- XGBoost hyperparameter tuning -----
xgb_base = XGBClassifier(
    objective="binary:logistic",
    eval_metric="logloss",
    use_label_encoder=False,
    random_state=42
)

xgb_param_dist = {
    "n_estimators": [100, 200, 300, 400],
    "max_depth": [3, 4, 5, 6],
    "learning_rate": [0.01, 0.05, 0.1, 0.2],
    "subsample": [0.7, 0.8, 1.0],
    "colsample_bytree": [0.7, 0.8, 1.0],
    "gamma": [0, 0.1, 0.2],
    "reg_lambda": [1, 5, 10],
}

xgb_random_search = RandomizedSearchCV(
    estimator=xgb_base,
    param_distributions=xgb_param_dist,
    n_iter=30,
    scoring="f1",
    cv=5,
    n_jobs=-1,
    random_state=42,
    verbose=1,
)

xgb_random_search.fit(X_train_scaled_Standard, y_train)

print("Best XGBoost params:")
print(xgb_random_search.best_params_)

best_xgb = xgb_random_search.best_estimator_

# Evaluate on test set
y_pred_xgb = best_xgb.predict(X_test_scaled_Standard)

print("\n=== Tuned XGBoost - Classification Report ===")
print(classification_report(y_test, y_pred_xgb, target_names=target_names))

cm_xgb = confusion_matrix(y_test, y_pred_xgb)
cm_xgb_df = pd.DataFrame(
    cm_xgb,
    index=[f"Actual {c}" for c in target_names],
    columns=[f"Pred {c}" for c in target_names]
)

print("\nConfusion Matrix (XGBoost, labeled):")
print(cm_xgb_df)

plt.figure(figsize=(5, 4))
sns.heatmap(cm_xgb_df, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix - Tuned XGBoost")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.tight_layout()
plt.show()
```
